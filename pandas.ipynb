# Missing values

# What do you mean by missing values in a dataset? How can you identify it and handle it using pandas?

Missing values refer to data that are absent, unavailable, or not recorded in a dataset. In Python Pandas, missing values are commonly represented by NaN (Not a Number), None, or pd.NA.

# Identify missing values

```python

print(df.isnull().sum())
```

# Handle missing values


## 2. How to Handle Missing Values Using Pandas

There are several methods to handle missing values in Pandas.

| Method | Pandas Command | Explanation |
|---|---|---|
| Remove rows | `df.dropna()` | Removes rows containing missing values |
| Remove columns | `df.dropna(axis=1)` | Removes columns containing missing values |
| Replace with mean | `df["Age"].fillna(df["Age"].mean())` | Fills missing ages with the average age |
| Replace with median | `df["Age"].fillna(df["Age"].median())` | Fills missing ages with the median |
| Replace with mode | `df["Name"].fillna(df["Name"].mode()[0])` | Fills missing names with the most frequent value |


# How to replace missing ages with the median

```python
# Check missing ages before cleaning
print(df["Age"].isnull().sum())

# Replace missing ages with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Check missing ages after cleaning
print(df["Age"].isnull().sum())
```

This is called median imputation.

# Important interview questions

Q1. What is the difference between isnull() and isna()?
Both perform the same operation in Pandas. They identify missing values.

Q2. What is the difference between isnull() and dropna()?
isnull() identifies missing values, whereas dropna() removes rows or columns containing missing values, depending on its parameters.

Q3. Why is .sum() used after .isnull()?
Because .isnull() returns Boolean values, and .sum() counts the True values to determine how many observations are missing.

Q4. Why is identifying missing values important in machine learning?
Many machine learning algorithms cannot directly handle missing numerical values. We therefore need to identify them and decide whether to remove, replace, or otherwise handle them before training a model.

Remember:
df.isnull() → Find missing values

df.isnull().sum() → Count missing values column-wise

df.fillna() → Replace missing values

df.dropna() → Remove missing values

# Use merge() and concat()

```python

# Merge two DataFrames
passengers = df[["PassengerId", "Name"]]
tickets = df[["PassengerId", "Fare"]]

merged = pd.merge(
    passengers,
    tickets,
    on="PassengerId"
)

print(merged.head())

# Concatenate rows
first = df.iloc[:5]
second = df.iloc[5:10]

combined = pd.concat(
    [first, second],
    ignore_index=True
)

print(combined)
```


## Difference between `pd.merge()` and `pd.concat()`

| Feature | `pd.merge()` | `pd.concat()` |
|---|---|---|
| Main purpose | Combine data using common keys | Stack DataFrames |
| Matching column required? | Usually yes | No |
| Default operation | Inner join | Vertical stacking |
| SQL equivalent | JOIN | Similar to UNION ALL when stacking rows |
| Titanic example | Combine passenger names with fares | Combine first 5 and next 5 passengers |
| Common application | Combining customer and transaction datasets | Combining monthly or yearly datasets |
