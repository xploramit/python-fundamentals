# Matplotlib - Data Visualization

Matplotlib allows us to create line graphs, bar charts, histograms, 

scatter plots, pie charts, and other visualizations.
For the following examples, use the original df.

Important Matplotlib commands

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv("titanic.csv")

```

# Histogram of passenger ages

```python
plt.hist(
    df["Age"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

# Bar chart of survival counts

```python

counts = df["Survived"].value_counts().sort_index()

plt.bar(
    ["Did Not Survive", "Survived"],
    counts.values
)

plt.title("Titanic Survival Counts")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.show()
```


| Code | Explanation |
|---|---|
| `df["Survived"]` | Selects the Survived column from the Titanic dataset. |
| `.value_counts()` | Counts passengers in each category (0 = Did Not Survive, 1 = Survived). |
| `.sort_index()` | Arranges categories in ascending order (0 first, then 1). |
| `plt.bar()` | Creates a bar chart. |
| `["Did Not Survive", "Survived"]` | Provides labels for the two bars on the X-axis. |
| `counts.values` | Provides the number of passengers in each category (bar heights). |
| `plt.title()` | Adds a title to the chart. |
| `plt.xlabel()` | Labels the X-axis. |
| `plt.ylabel()` | Labels the Y-axis. |
| `plt.show()` | Displays the chart. |


```python
counts = df["Survived"].value_counts()

plt.bar(
    ["Did Not Survive", "Survived"],
    counts.values
)

plt.title("Titanic Survival Counts")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.show()
```

# Bar chart of survival rate by sex

```python

rates = df.groupby("Sex")["Survived"].mean() * 100

plt.bar(rates.index, rates.values)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.show()
```

# Scatter plot of age and fare

```python
plt.scatter(
    df["Age"],
    df["Fare"],
)

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

# Boxplot of fares

```python

plt.boxplot(df["Fare"].dropna())

plt.title("Distribution of Ticket Fares")
plt.ylabel("Fare")
plt.show()
```

# A box plot helps identify the median, quartiles, spread, and potential outliers.

# Line graph of average fare by passenger class

```python

avg_fare = df.groupby("Pclass")["Fare"].mean()

plt.plot(
    avg_fare.index,
    avg_fare.values,
    marker="o"
)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.xticks([1, 2, 3])
plt.show()
```

# Multiple plots in one figure

```python

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(
    df["Age"].dropna(),
    bins=20
)
axes[0].set_title("Age Distribution")

axes[1].hist(
    df["Fare"].dropna(),
    bins=20
)
axes[1].set_title("Fare Distribution")

plt.tight_layout()
plt.show()
```

# Save a graph

```python

plt.hist(df["Age"].dropna(), bins=20)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.savefig("age_distribution.png", dpi=300)
plt.show()
```

# What is matplotlib in python? what is its importance and give any four types of graphs that can be created using it.

Matplotlib is a Python library used for data visualization. It helps represent numerical and 

statistical data through graphs and charts, making it easier to understand patterns, trends, and relationships in datasets.

# Importance of Matplotlib

1. **Data Visualization**: Converts complex numerical data into easy-to-understand graphs.

2. **Identifying Trends**: Helps identify trends and patterns in datasets.

3. **Comparing Data**: Makes it easier to compare different categories or variables.

4. **Understanding Relationships**: Helps examine relationships between two or more variables.

5. **Decision-Making**: Supports data analysis, research, and informed decision-making.

# Four Types of Graphs in Matplotlib


| Graph | Purpose | Python Command |
|---|---|---|
| **Line Graph** | Shows trends or changes over time, such as GDP growth. | `plt.plot()` |
| **Bar Chart** | Compares values across categories, such as sales of different products. | `plt.bar()` |
| **Histogram** | Shows the frequency distribution of numerical data, such as household income. | `plt.hist()` |
| **Scatter Plot** | Shows the relationship between two numerical variables, such as income and consumption. | `plt.scatter()` |


```python

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# Create a line graph
plt.plot(x, y, marker="o")

plt.title("Simple Line Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```
