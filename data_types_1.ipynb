# What is string interpolation in Python? (i.e., str.format(), f” ” )

"Suppose Amazon has 10 crore customers. Does Amazon write 10 crore different messages manually?"

String interpolation means inserting the value of a variable or expression inside a string.

Why do we need string interpolation?

Suppose you are building a college application.

```python
name = "Rahul"
marks = 85
```

You want Python to display:

Hello Rahul, you scored 85 marks.

Tomorrow another student logs in:

name = "Priya"

marks = 92

Now the program should automatically display:

Hello Priya, you scored 92 marks.

We cannot manually type a separate message for every student. Instead:

```python
name = "Mahima"
marks = 95
gender = "female"

print(f"Hello {name}, you scored {marks} marks and you are {gender}.")
```

The sentence remains fixed, but the data inside it changes dynamically.

That is the practical purpose of string interpolation.

# TUPLES

A tuple is used to store multiple values in a single variable. A tuple can contain values of the same type or different data types. 

Unlike a list, a tuple is immutable, meaning its elements cannot normally be changed after the tuple has been created.

```python
student = ("Zella", 21, "The Assam Royal Global University")
print(student)
```

# 1. Important Features of a Tuple

A tuple:

Can store multiple elements.

Can contain homogeneous or heterogeneous data.

Is ordered — insertion order is preserved.

Is immutable — elements cannot normally be added, removed or replaced.

Allows duplicate values.

Supports positive and negative indexing.

Supports slicing.

Is usually written using parentheses ().

Can also be created without parentheses.

Requires a comma when creating a tuple containing only one element.

Can be nested inside another tuple.

# Easy way to remember

List = changeable []

Tuple = fixed ()

# 6. A Very Interesting Example – Function Returning Multiple Values

This is one of the best reasons to teach tuples.

Suppose Amazon-like billing software calculates both total price and discount:

```python
def calculate_bill(price):
    discount = price * 0.10
    final_price = price - discount
    return discount, final_price

result = calculate_bill(50000)

print(result)
```

Python has returned the values together as a tuple.

We can unpack them immediately:

```python
discount, final_price = calculate_bill(50000)

print(f"Discount = ₹{discount}")
print(f"Final Price = ₹{final_price}")
```
