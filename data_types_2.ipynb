# 1. What is a List?

A list is a data structure in Python used to store multiple values in a single variable.

For example, suppose we want to store the names of five students:

A Python list is an ordered, mutable collection that can contain duplicate values and different data types. Its elements can be accessed using indexing and slicing.

```python
students = ["Amit", "Nishant", "Zella", "Bikash", "Haflong"]

print(students)
```

# 2. Important Features of a List


A. A list can contain homogeneous elements

Homogeneous means all elements are of the same data type.

```python
marks = [75, 82, 68, 90, 85]

print(marks)
```

Here, all elements are integers.

Another example:

```python
students = ["Amit", "Nishant", "Zella", "Bikash"]

print(students)
```

Here, all elements are strings.

# B. A list can contain heterogeneous elements

Heterogeneous means elements can be of different data types.

Example:

```python
student = ["Amit", 25, 85.5, True]

print(student)
```

Here:

"Amit" → string

25 → integer

85.5 → float

True → Boolean

So a Python list does not require all elements to have the same data type.

# 3. Lists Allow Duplicate Values

A list can contain the same value more than once.

Example:

```python
places = ["Haflong", "Guwahati", "Haflong", "Shillong"]

print(places)
```

# 4. Insertion Order is Preserved

Python maintains the order in which elements are inserted into a list.

For example:

```python
students = ["Amit", "Nishant", "Zella", "Bikash"]

print(students)
```

Amit was entered first, so Amit remains first. Nishant was entered second, so Nishant remains second, and so on.

# 7. List Indexing

Indexing means accessing one particular element from a list.

```python
print(students[0])
```

```python
print(students[-2])
```

# 8. List Slicing

Slicing means extracting multiple elements from a list.

The colon : is called the slicing operator.

Basic syntax:

# list[start:stop]

```python
print(students[0:3])
```

# 12. Creating a List Using range()



range() is useful when we want to generate a sequence of numbers.

Usually we convert it to a list like this:`

# 14. Lists are Mutable

This is one of the most important characteristics of lists.

Mutable means changeable.

```python
students = ["Amit", "Nishant", "Zella"]
```

Now suppose Nishant is to be replaced by Bikash.

```python
students[1] = "Bikash"

print(students)
```

# 15. List Concatenation

Concatenation means joining two or more lists together.

The + operator is used.

Suppose:

```python
list1 = ["Amit", "Nishant"]
list2 = ["Zella", "Bikash"]

students = list1 + list2

print(students)
```

# 16. List Multiplication / List Repetition

Python allows us to repeat a list using the * operator.

Suppose:

```python
students = ["Amit", "Zella"]

result = students * 3

print(result)
```

# Important List Functions and Methods in Python

The important list methods are len(), count(), index(), append(), extend(), insert(), remove(), pop(), reverse(), copy(), clear(), max(), min() and sort().

# 1. len() — Count items

len() tells us how many elements are present in a list.

Instagram example: How many people liked a post?

```python
likes = ["Amit", "Nishant", "Zella", "Bikash"]

print(len(likes))
```

# 2. append() — Add one item

append() adds one new element at the end of a list.

```python
cart = ["Laptop", "Mouse"]

cart.append("Keyboard")

print(cart)
```

# 3. extend() — Add multiple items

extend() adds multiple elements to a list.

```python
cart = ["Laptop"]

cart.extend(["Mouse", "Keyboard", "Charger"])

print(cart)
```

# Remember:

# append() → ONE item

# extend() → MANY items

# 4. insert() — Add at a particular position

insert() adds an element at a specific index.

```python
students = ["Amit", "Zella", "Bikash"]

students.insert(1, "Nishant")

print(students)
```

# 5. remove() — Remove by value

Suppose your Swiggy cart is:

```python
cart = ["Pizza", "Burger", "Biryani"]

cart.remove("Burger")

print(cart)
```

remove() removes the specified value; with duplicates, it removes the first occurrence.

# 6. pop() — Remove by index

```python
cart = ["Pizza", "Burger", "Biryani"]

cart.pop(2)

print(cart)
```

# remove("Burger") → remove by VALUE

# pop(1) → remove by POSITION

# 7. sort() — Arrange items

Amazon example: Price Low to High.

```python
prices = [999, 299, 1499, 499]

prices.sort()

print(prices)
```

# For High to Low:

```python
prices.sort(reverse=True)
print(prices)
```

# 8. max() and min() — Highest and lowest

Suppose Swiggy has these delivery times:

```python
time = [20, 45, 15, 30]

print(max(time))
print(min(time))
```
