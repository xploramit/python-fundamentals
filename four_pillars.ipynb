# The Four Main Pillars of OOP

Once clear with class, object, attributes and methods, introduce the famous four concepts:

# Encapsulation → Abstraction → Inheritance → Polymorphism

“We already know Class, Object, Attribute and Method. Now we will see the four important ideas of OOP using the same Amazon example.”

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.price)

    def apply_discount(self, discount):
        discount_amount = self.price * discount / 100
        final_price = self.price - discount_amount

        print("Original Price:", self.price)
        print("Discount:", discount, "%")
        print("Final Price:", final_price)


p1 = Product("iPhone 17", 79999)
p2 = Product("Samsung Galaxy S26", 69999)
```

# Encapsulation

Encapsulation means keeping data and the methods that work on that data together inside a class, 
and controlling how sensitive data is accessed or changed.

Our existing Product class already shows the first part:

Suppose Amazon does not want anyone to directly change a product's price.

We can write:

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.__price)

    def apply_discount(self, discount):
        final_price = self.__price - (self.__price * discount / 100)
        print("Final Price:", final_price)
```

Notice:

# self.__price

The double underscore __ makes the attribute harder to access directly from outside the class through Python's name-mangling mechanism.
So the idea is:

Amazon real-life idea
A customer should be able to:

View Price

Apply Coupon

Buy Product

# But a customer should not be able to directly say:
# Change iPhone price from ₹79,999 to ₹999

# Encapsulation = Keep related data and methods together and control access to the data.

# Encapsulation → PROTECT

# Abstraction means hiding unnecessary internal details and showing the user only what is necessary.

We can demonstrate this with a simple method:

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def buy_now(self):
        print("Checking stock...")
        print("Processing payment...")
        print("Creating order...")
        print("Order successful!")
```

buy_now()

That is abstraction.

# Inheritance

Now suppose Amazon sells different types of products:

All products have some common features:

name

price

display_details()

Instead of writing these again and again, we can create a general Product class.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.price)
```

# Now create an Electronics class:

```python
class Electronics(Product):

    def show_warranty(self):
        print("Warranty: 1 Year")
```

Notice:

class Electronics(Product):


This means: Electronics inherits from Product.

```python
p1 = Electronics("iPhone 17", 79999)
```

```python
p1.display_details()
```

But wait!
We never wrote display_details() inside Electronics.

It received it from: Product

This is inheritance.

```python
p1.show_warranty()
```

```python
p1.display_details()
p1.show_warranty()
```

Inheritance means one class can acquire and reuse the properties and methods of another class.

# Inheritance → REUSE

| Type | Remember It As | Example |
|---|---|---|
| **Single** | One Parent → One Child | `Product → Electronics` |
| **Multiple** | Many Parents → One Child | `Product + Discount → Smartphone` |
| **Multilevel** | Grandparent → Parent → Child | `Product → Electronics → Smartphone` |
| **Hierarchical** | One Parent → Many Children | `Product → Electronics, Clothing, Books` |
| **Hybrid** | Combination | Multiple + Hierarchical |

# Polymorphism

POLY = Many

MORPH = Forms

Therefore: Polymorphism = Many Forms

Suppose Amazon sells:

Electronics, Clothing, Books

All products may have:

delivery()


But delivery may work differently.
For electronics:

# For electronics:

```python
class Electronics:

    def delivery(self):
        print("Secure electronic delivery")
```

# For clothing:

```python
class Clothing:

    def delivery(self):
        print("Standard clothing delivery")
```

Now:

```python
iphone = Electronics()

shirt = Clothing()
```

Both use the same method name, But the output is different for the below input. 

```python
iphone.delivery()
shirt.delivery()
```

# An Even Easier Amazon Example for Polymorphism

UPI
Credit Card
Cash on Delivery

All perform:

pay()

But they perform it differently.

```python
class UPI:

    def pay(self):
        print("Payment made using UPI")


class CreditCard:

    def pay(self):
        print("Payment made using Credit Card")


class CashOnDelivery:

    def pay(self):
        print("Payment will be made at delivery")
```

```python
p1 = UPI()
p2 = CreditCard()
p3 = CashOnDelivery()

p1.pay()
p2.pay()
p3.pay()
```

“Is the method name changing?”

No.
It is always:

pay()

But its behaviour changes depending on the object. That's polymorphism.

# Polymorphism means the same method/action can behave differently for different objects.
