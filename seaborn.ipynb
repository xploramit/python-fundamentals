# Seaborn - Statistical Data Visualization
Seaborn is built on Matplotlib and makes it easier to create attractive statistical visualizations.

# Important Seaborn commands

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv("titanic.csv")
```

# Count plot of survival status

```python
sns.countplot(
    data=df,
    x="Survived"
)

plt.title("Titanic Survival Counts")
plt.show()
```

# Survival counts by sex

```python
sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival Counts by Sex")
plt.show()
```

# Survival rate by passenger class

```python

sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    errorbar=None
)

plt.title("Survival Rate by Passenger Class")
plt.ylabel("Survival Proportion")
plt.show()
```

Because Survived is coded as 0 and 1, the mean displayed by sns.barplot() represents the survival proportion.

# Histogram with density curve

```python

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Age Distribution with Density Curve")
plt.show()
```

# Boxplot of fares by class

```python

sns.boxplot(
    data=df,
    x="Pclass",
    y="Fare"
)

plt.title("Fare Distribution by Passenger Class")
plt.show()
```

# Scatter plot with survival categories

```python

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age, Fare and Survival")
plt.show()
```

# Correlation heatmap

```python

numeric = df[
    ["Survived", "Pclass", "Age",
     "SibSp", "Parch", "Fare"]
]

corr = numeric.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

The heatmap displays pairwise Pearson correlations. Correlation involving the numerical codes of Pclass should be interpreted carefully because passenger class is an ordinal category.

# Pair plot

```python

sns.pairplot(
    df[["Age", "Fare", "Pclass", "Survived"]],
    hue="Survived"
)

plt.show()
```

This displays relationships between several numerical variables simultaneously.

# Violin plot

```python

sns.violinplot(
    data=df,
    x="Survived",
    y="Age"
)

plt.title("Age Distribution by Survival Status")
plt.show()
```

# Regression plot

```python

sns.regplot(
    data=df,
    x="Age",
    y="Fare",
    scatter_kws={"alpha": 0.3}
)

plt.title("Age and Fare with Regression Line")
plt.show()
```

This adds a fitted linear regression line to the scatter plot. It provides a natural introduction to linear regression.

# Survival rate heatmap by sex and class

```python

survival_table = df.pivot_table(
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
) * 100

sns.heatmap(
    survival_table,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("Survival Rate (%) by Sex and Class")
plt.show()
```

This is one of the most useful graphs for demonstrating how survival patterns differed across passenger groups.

# What is a correlation heatmap? How can you create it using seaborn?

A correlation heatmap is a graphical representation of the correlation coefficients 

between numerical variables in a dataset. It uses different colours to show the strength 

and direction of relationships between variables.

**Correlation coefficients range from −1 to +1**:
- +1: Perfect positive correlation.
  
- −1: Perfect negative correlation.

  
- 0: No linear correlation.

  
Python Program Using Seaborn (Titanic Dataset)

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
df = pd.read_csv("titanic.csv")

# Select numerical variables
data = df[["Survived", "Pclass", "Age", "Fare"]]

# Calculate correlation matrix
corr = data.corr()

# Create correlation heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Titanic Correlation Heatmap")
plt.show()
```

# Explanation of Commands


| **Command** | **Purpose** |
|---|---|
| `df.corr()` | Calculates correlation coefficients between numerical variables. |
| `sns.heatmap()` | Creates a correlation heatmap. |
| `annot=True` | Displays correlation values inside the cells. |
| `cmap="coolwarm"` | Uses a colour scale to represent positive and negative correlations. |
| `plt.title()` | Adds a title to the graph. |
| `plt.show()` | Displays the heatmap. |


- **Survived and Pclass**: Negative correlation (approximately −0.34), indicating that

  passengers in lower-numbered classes tended to have higher survival rates.
  
- **Survived and Far**e: Positive correlation (approximately +0.26), indicating that

  passengers paying higher fares tended to have higher survival rates.

  
- **Age and Fare**: Weak positive correlation (approximately +0.10).


**Conclusion**: A correlation heatmap helps identify positive and negative linear relationships 

between variables, making it useful for exploratory data analysis and selecting 

variables for 
further statistical analysis or machine learning.
