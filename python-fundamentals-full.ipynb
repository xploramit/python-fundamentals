# Comparison Operators in Python

Comparison operators are used to compare two values. After comparing the values, Python gives the answer as either:

True – if the comparison is correct
False – if the comparison is not correct

For example, suppose:

```python
x = 5
y = 10
```

Now we can compare x and y using different comparison operators.

| Operator | Meaning                  | Example  | Result  |
| -------- | ------------------------ | -------- | ------- |
| `==`     | Equal to                 | `x == y` | `False` |
| `!=`     | Not equal to             | `x != y` | `True`  |
| `>`      | Greater than             | `x > y`  | `False` |
| `<`      | Less than                | `x < y`  | `True`  |
| `>=`     | Greater than or equal to | `x >= y` | `False` |
| `<=`     | Less than or equal to    | `x <= y` | `True`  |


```python
x = 5
y = 10
```

```python
print(x == y)   # False
print(x != y)   # True
print(x > y)    # False
print(x < y)    # True
print(x >= y)   # False
print(x <= y)   # True
```

One important point: = and == are different in Python.

```python
x = 5       # Assigns 5 to x
x == 5      # Checks whether x is equal to 5
```

So, = means assignment, whereas == means comparison.

# Ternary Operator

The ternary operator is simply a short way of writing a simple if-else condition in one line.

Suppose:

```python
x = 10
y = 20
```

We want to find which number is greater.

Normally, we could write:

```python
if x > y:
    result = x
else:
    result = y
```

But the ternary operator lets us write this in one line:

```python
result = x if x > y else y
```

How do we read this?

Read it almost like English:

Take x if x > y; otherwise take y.

Since:

x = 10
y = 20

the condition:

x > y

is:

False

Therefore Python chooses y.

```python
print(result)
```

# For Loop in Python

Why do we need a loop?

Suppose we want to print "Hello" five times.

Without a loop, we would write:

print("Hello")

print("Hello")

print("Hello")

print("Hello")

print("Hello")

This is repetitive.

Instead, we can use a loop:

```python
for i in range(5):
    print("Hello")
```

range(5) generates these numbers:

0, 1, 2, 3, 4

So the loop runs 5 times.

Here is what happens step by step:

| Loop | Value of `i` | What Python does |
| ---: | -----------: | ---------------- |
|    1 |            0 | Prints `Hello`   |
|    2 |            1 | Prints `Hello`   |
|    3 |            2 | Prints `Hello`   |
|    4 |            3 | Prints `Hello`   |
|    5 |            4 | Prints `Hello`   |


So, a loop is used to repeat a block of code multiple times.

A for loop is used to repeat a set of statements for each item in a sequence.

Basic Syntax

for variable in sequence:

    statement

```python
for i in range(5):
    print(i)
```

# What is range()?

The range() function is used to generate a sequence of integers.

It is commonly used with a for loop when we want to repeat something a specific number of times.

Syntax
range(start, stop, step)

It can take up to three arguments:

| Argument | Meaning                               |           Default |
| -------- | ------------------------------------- | ----------------: |
| `start`  | Where the sequence starts             |               `0` |
| `stop`   | Where the sequence stops              | Must be specified |
| `step`   | Difference between consecutive values |               `1` |

Important Rule

The stop value is not included in the sequence.

range(stop)

If we provide only one value:

range(10)

Python understands it as:

range(0, 10, 1)

# While Loop in Python

A while loop is used to repeat a block of code as long as a given condition is True.

In simple words:

While this condition is true, keep doing the work.

# Syntax

```python
while condition:
    statement1
    statement2
```

Python first checks the condition.

If the condition is True → the statements inside the loop are executed.

Python checks the condition again.

The process continues as long as the condition remains True.

When the condition becomes False → the loop stops.

```python
num = 1

while num <= 5:
    print("The count is:", num)
    num = num + 1
```

# Countdown Using while

```python
i = 5

while i >= 1:
    print(i)
    i -= 1
```

# for Loop vs while Loop

| `for` loop                                         | `while` loop                                        |
| -------------------------------------------------- | --------------------------------------------------- |
| Usually used when we know how many times to repeat | Usually used when repetition depends on a condition |
| Often used with `range()`                          | Uses a condition                                    |
| Automatically moves to the next item               | We usually need to update the variable ourselves    |


# Transfer Statements

Transfer statements are used to transfer the program control from one location to another location.
We have different types of transfer statements,
1. break
2. continue
3. pass
4. return
5. assert

# 1. break Statement

The break statement is used to stop a loop immediately.

Suppose we want to print numbers from 1 onwards but stop when we reach 5.

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

### How `break` Works Step by Step

| Value of `i` | Is `i == 5`? | Action |
|---|---|---|
| `i = 1` | No | Print `1` |
| `i = 2` | No | Print `2` |
| `i = 3` | No | Print `3` |
| `i = 4` | No | Print `4` |
| `i = 5` | Yes | **BREAK → Loop stops** |

# 2. continue Statement

The continue statement is used to skip the current iteration and move to the next iteration.

Example:

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

Notice that 3 is missing.

What happened?

i = 1 → Print 1

i = 2 → Print 2

i = 3 → CONTINUE → Skip print(i)

i = 4 → Print 4

i = 5 → Print 5

The loop does not stop. It only skips the current iteration.

Meaning

continue = Skip this one and continue with the next.

# break vs continue

| Statement  | What does it do?                 |
| ---------- | -------------------------------- |
| `break`    | Stops the entire loop            |
| `continue` | Skips only the current iteration |


# 3. pass Statement

The pass statement means:

Do nothing.

It is used when Python requires a statement, but we don't want to write the code yet.

```python
age = 20

if age >= 18:
    pass
```

```python
# Python will not produce an error.
```

Suppose you are developing a program:

```python
def calculate_tax():
    pass
```

We are telling Python:

"I will write the tax calculation later. For now, leave this function empty."

Easy meaning

pass = Do nothing and move on.

# 4. return Statement

return is mainly used inside a function.

It sends a result back from the function and ends that function's execution.

For example:

```python
def add(a, b):
    result = a + b
    return result

x = add(10, 20)

print(x)
```

return = Send the result back from a function.

We will understand return more clearly when you start functions in Python.

# 5. assert Statement

assert is used to check whether something that we expect to be true is actually true.

For example:

```python
age = 20

assert age >= 18
```

Since:

20 >= 18 → True

nothing happens.

But:

```python
age = 15

assert age >= 18
```

will produce an:

AssertionError

We can also provide a message:

```python
age = 15

assert age >= 18, "Age must be 18 or above"
```

Easy meaning

assert = Check that an assumption is true; otherwise raise an error.

It is mainly useful for debugging and testing assumptions in code, not as a general replacement for normal user-input validation.

| `pass`     | **Do nothing**          | Empty blocks/placeholders |
| `return`   | **Send back** a result  | Functions                 |
| `assert`   | **Check** an assumption | Testing/debugging         |
