# SET – Data Type in Python

# 1. What is a Set?

A set in Python is an unordered collection of unique elements. 

It does not allow duplicate values and is useful for membership testing and 

operations such as union, intersection and difference.

A set is a collection of unique elements. It does not allow duplicate values, and unlike a list, you should not rely on its element order.

Instagram example

Suppose these users liked a post:

```python
likes = {"Amit", "Zella", "Bikash", "Amit"}

print(likes)

len(likes)

type(likes)
```

The set will contain "Amit" only once.

Conceptually:

{'Amit', 'Zella', 'Bikash'}

Even though Amit was entered twice.

# Why is this useful?


Imagine Instagram wants to know:

How many unique users liked a post?

A set is useful because duplicates are automatically removed.

# 2. Creating a Set

We normally create a set using curly brackets { }.

```python
customers = {"Amit", "Nishant", "Zella", "Bikash"}

print(customers)
```

Or using set():

```python
customers = set(["Amit", "Nishant", "Zella"])
print(customers)
```

# 3. Sets Do Not Support Indexing

With a list, we can write:

students[0]

But we cannot do this with a set.

```python
students = {"Amit", "Zella", "Bikash"}

students[0]    # Error
```

Why?

Because sets are unordered collections, so there is no meaningful fixed index 0, 1, 2, etc.

# 4. add() – Add ONE Element

add() adds one new element to a set.

# Facebook example

Suppose these are members of a Facebook group:

```python
members = {"Amit", "Nishant", "Zella"}

members.add("Gungunnnnnnnnnnnnnn")

print(members)
```

# 5. update() – Add MANY Elements

update() is used to add multiple elements.

```python
members = {"Amit", "Nishant"}

members.update(["Zella", "Bikash"])

print(members)
```

So remember:

# add() → ONE

# update() → MANY

# 6. remove() and discard() – Remove Elements

Both are used to remove an element from a set.

remove()

```python
customers = {"Amit", "Nishant", "Ngiplo"}

customers.remove("Nishant")
```

```python
print(customers)
```

But if you try:

customers.remove("Bikash")

when Bikash doesn't exist, Python gives a KeyError.

```python
customers.remove("Bikash")
```

# discard()

```python
customers.discard("Bikash")
```

If Bikash isn't present, Python simply does nothing.

This is the main distinction highlighted in your material.

# Difference

| Method      | If element doesn't exist |
| ----------- | ------------------------ |
| `remove()`  | ❌ Error                  |
| `discard()` | ✅ No error               |


# 7. Membership – Is Something Present?

Sets are especially useful for checking whether something exists in a collection.

# Zomato example

Suppose:

```python
restaurants = {"KFC", "Dominos", "Barbeque Nation"}

print("KFC" in restaurants)
```

```python
print("Pizza Hut" in restaurants)
```

Think of it as:

# Zomato asking: “Is this restaurant in my available set of restaurants?”

# 8. Union – Combine Everything

Union means all unique elements from both sets.

Imagine:

```python
instagram = {"Amit", "Nishant", "Zella"}

facebook = {"Amit", "Bikash", "Zella"}
```

Notice that Amit and Zella are not repeated.

Union = Everything from both sets

# We want everyone who uses Instagram OR Facebook or both:

```python
print(instagram.union(facebook))
```

```python
instagram | facebook
```

# 9. Intersection – Find Common Elements

Intersection means elements that are present in BOTH sets.

Using the same example:

```python
instagram = {"Amit", "Nishant", "Zella"}
facebook = {"Amit", "Bikash", "Zella"}
```

```python
print(instagram.intersection(facebook))
```

```python
instagram & facebook
```

Because Amit and Zella use both platforms.

# Intersection = COMMON

# 10. Difference – What's in One but Not the Other?

Difference gives elements present in the first set but not in the second set.

```python
instagram = {"Amit", "Nishant", "Zella"}
facebook = {"Amit", "Bikash", "Zella"}

print(instagram - facebook)
```

```python
instagram - facebook
```

Meaning:

Nishant is in the Instagram set but not in the Facebook set.

If we reverse it:

```python
facebook - instagram
```

# So order matters in difference.

# Most Important Part: Union vs Intersection vs Difference

# Using one simple example throughout the class:

```python
swiggy = {"Amit", "Nishant", "Zella"}
zomato = {"Amit", "Bikash", "Zella"}
```

Union |

Who uses Swiggy OR Zomato or both?

```python
swiggy | zomato
```

# Intersection &

Who uses BOTH Swiggy and Zomato?

```python
swiggy & zomato
```

# Difference -

# Who uses Swiggy but not Zomato?

```python
swiggy - zomato
```

# These mathematical operations are one of the main purposes of sets in Python.

# Dictionary in Python — Short Notes with Real-Life Examples

# 1. What is a Dictionary?

A dictionary stores information in key : value pairs.

Think about an Instagram profile:

```python
profile = {
    "username": "amit_kumar",
    "followers": 5000,
    "verified": True
}
```

| Key           | Value          |
| ------------- | -------------- |
| `"username"`  | `"amit_kumar"` |
| `"followers"` | `5000`         |
| `"verified"`  | `True`         |


So, simply:

Dictionary = Key : Value

Just like a real dictionary:

Word → Meaning

Python dictionary:

Key → Value

# 2. Creating a Dictionary

Dictionaries are commonly created using { }.

Amazon example

```python
product = {
    "name": "Laptop",
    "price": 55000,
    "rating": 4.5
}

print(product)
```

A dictionary can contain different types of information about the same object.

# 3. Keys Must Be Unique

A dictionary cannot have duplicate keys. If the same key is assigned again, its value is updated.

For example:

```python
product = {
    "p
    
    rice": 50000
}

product["price"] = 45000

print(product)
```

# Think of an Amazon price update:

# Laptop price was ₹50,000 → Amazon changes it to ₹45,000.

# The key "price" remains the same; its value changes.

# 4. Accessing Values Using Keys

Unlike a list, we normally access dictionary values using their keys.

# Instagram example

```python
profile = {
    "username": "amit_kumar",
    "followers": 5000,
    "posts": 120
}

print(profile["followers"])
```

We are basically asking:

What is the value associated with "followers"?

# 5. Adding New Information

We can easily add a new key-value pair.

LinkedIn example

```python
profile = {
    "name": "Amit",
    "job": "Professor"
}

profile["city"] = "Guwahati"

print(profile)
```

# Important Dictionary Methods

# 6. keys() — Get All Keys

Imagine an Amazon product:

```python
product = {
    "name": "iPhone",
    "price": 70000,
    "rating": 4.6
}

print(product.keys())
```

This gives the dictionary's keys:

name, price, rating

keys() returns the keys in the dictionary.

# 7. values() — Get All Values

```python
print(product.values())
```

remember:

keys() → labels

values() → information associated with those labels

# 8. items() — Get Keys AND Values

```python
print(product.items())
```

```python
print(product.items())
```

This gives all the key-value pairs.

Think:

keys() → only keys
    
values() → only values
    
items() → both

# 9. get() — Safely Get a Value
Zomato example

```python
restaurant = {
    "name": "Food Corner",
    "rating": 4.5,
    "delivery": 30
}

print(restaurant.get("rating"))
```

get() retrieves the value associated with a specified key.

A useful advantage is that:

```python
print(restaurant.get("discount"))
```

returns None if "discount" doesn't exist, rather than giving a KeyError.

# 10. pop() — Remove a Key-Value Pair

Suppose a Facebook profile contains:

```python
profile = {
    "name": "Amit",
    "age": 30,
    "city": "Guwahati"
}
```

# Remove "age":

```python
profile.pop("age")

print(profile)
```

# pop() removes a specified key-value pair and returns its value.

# 11. update() — Add or Update Information

Suppose a Swiggy restaurant has:

```python
restaurant = {
    "name": "Food Hub",
    "rating": 4.2
}
```

New information arrives:

```python
restaurant.update({
    "rating": 4.5,
    "delivery": 25
})

print(restaurant)
```

# Notice:

"rating" already existed → updated

"delivery" didn't exist → added

That's the important idea behind update().
