# What is NumPy?

NumPy stands for Numerical Python. It is a Python library used for 

performing numerical and mathematical calculations efficiently.

It is especially useful when working with large amounts of numerical data.

For example, an economist might use NumPy to calculate:
- Average household income
- Inflation and GDP growth rates
- Standard deviation of wages
- Correlation between income and consumption
- Matrix operations used in econometrics
  
Why do we need NumPy when Python already has lists?
Consider the following example.

# Without NumPy

```python

income = [10000, 15000, 20000, 25000, 30000]

new_income = [x * 1.10 for x in income]

print(new_income)
```

# With NumPy

```python

import numpy as np

income = np.array([10000, 15000, 20000, 25000, 30000])

new_income = income * 1.10

print(new_income)
```

Both programs increase income by 10%. However, NumPy performs the 

calculation directly on the entire array without requiring an explicit Python loop.

# Interview question: Why is NumPy generally faster than Python lists?

NumPy stores homogeneous numerical data efficiently and performs many operations using optimized compiled code, 

reducing the overhead of Python-level loops.

# Array properties

```python

a = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])

print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
print(a.itemsize)
print(a.nbytes)
print(a.data)
```


| Command | Meaning | Result |
|---|---|---|
| `a.ndim` | Number of dimensions | 2 |
| `a.shape` | Rows and columns | (2, 4) |
| `a.size` | Number of elements | 8 |
| `a.dtype` | Data type | Platform-dependent integer type |
| `a.itemsize` | Bytes per element | Depends on dtype |
| `a.nbytes` | Total bytes occupied by elements | size × itemsize |
| `a.data` | Buffer exposing array data | Memory buffer |


Interview question: What is the difference between shape, size, and ndim?

Answer: shape tells us the size along each dimension, size tells us the total number of elements, and ndim tells us the number of dimensions.

# Mathematical operations using NumPy

# Addition, subtraction, multiplication and division

```python

import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

a * b performs element-wise multiplication, not matrix multiplication. For matrix multiplication, use a @ b or np.matmul() with compatible array shapes.

# Scalar operations

```python

a = np.array([10, 20, 30])

print(a + 5)
print(a - 5)
print(a * 5)
print(a / 5)
print(a ** 2)
```

# Statistical operations using NumPy


| Command | Statistical Purpose |
|---|---|
| `np.sum(a)` | Sum |
| `np.mean(a)` | Arithmetic mean |
| `np.median(a)` | Median |
| `np.std(a)` | Standard deviation |
| `np.var(a)` | Variance |
| `np.min(a)` | Minimum |
| `np.max(a)` | Maximum |
| `np.percentile(a, 25)` | 25th percentile |
| `np.quantile(a, 0.5)` | Median quantile |
| `np.corrcoef(a, b)` | Correlation matrix |
| `np.cov(a, b)` | Covariance matrix |


```python

import numpy as np

income = np.array([10000, 15000, 20000, 25000, 30000])

print("Total:", np.sum(income))
print("Mean:", np.mean(income))
print("Median:", np.median(income))
print("Standard Deviation:", np.std(income))
print("Variance:", np.var(income))
print("Minimum:", np.min(income))
print("Maximum:", np.max(income))
```

# Correlation between income and consumption

```python

income = np.array([10000, 15000, 20000, 25000, 30000])
consumption = np.array([8000, 11000, 15000, 18000, 22000])

correlation = np.corrcoef(income, consumption)

print(correlation)
print("Correlation coefficient:", correlation[0, 1])
```

# Importing real-world datasets and performing NumPy analysis

NumPy is used for performing mathematical and statistical operations on arrays.

We will extract columns from the Titanic dataset and perform numerical calculations.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv("titanic.csv")

# Display first five rows
print(df.head())
```

```python

age = df["Age"].to_numpy()
fare = df["Fare"].to_numpy()
survived = df["Survived"].to_numpy()

print(age[:10])
print(fare[:10])
print(survived[:10])
```

# .to_numpy() converts a Pandas column into a NumPy array. [:10] selects the first 10 elements.

# Check array dimensions, shape, size and data type

```python

print("Dimensions:", age.ndim)
print("Shape:", age.shape)
print("Size:", age.size)
print("Data type:", age.dtype)
print("Bytes occupied:", age.nbytes)
```


| Command | Purpose |
|---|---|
| `ndim` | Number of dimensions |
| `shape` | Number of elements along each dimension |
| `size` | Total number of elements |
| `dtype` | Type of data |
| `nbytes` | Memory occupied by array elements |


# Calculate mean, median, minimum and maximum

```python
print("Mean Age:", np.nanmean(age))
print("Median Age:", np.nanmedian(age))
print("Youngest:", np.nanmin(age))
print("Oldest:", np.nanmax(age))
print("Average Fare:", np.mean(fare))
```

# Why nanmean() instead of mean()? Some passengers have missing ages (NaN). np.nanmean() ignores these missing values.

# Calculate standard deviation and variance

```python

print("Age SD:", np.nanstd(age))
print("Age Variance:", np.nanvar(age))

print("Fare SD:", np.std(fare))
print("Fare Variance:", np.var(fare))
```

# Standard deviation measures how dispersed observations are around their mean.

# Boolean indexing - filtering data
# Find passengers older than 60.

```python
elderly = age[age > 60]

print(elderly)
print("Number of elderly passengers:", len(elderly))
```

# Find passengers who paid fares greater than 100.

```python

high_fare = fare[fare > 100]

print(high_fare)
print("Number of passengers:", len(high_fare))
```

# Calculate survival rate

```python

survival_rate = np.mean(survived) * 100

print("Survival Rate:", survival_rate)
```

Because Survived contains only 0 and 1, its arithmetic mean represents the proportion of passengers who survived.

# Calculate survival rate by passenger class

```python

pclass = df["Pclass"].to_numpy()

for cls in [1, 2, 3]:
    rate = np.mean(survived[pclass == cls]) * 100
    print(f"Class {cls}: {rate:.2f}%")
```

Did first-class passengers have a higher survival rate than third-class passengers?

# Calculate correlation

```python

# Exclude missing ages
valid = ~np.isnan(age)

correlation = np.corrcoef(
    age[valid],
    fare[valid]
)[0, 1]

print("Age-Fare Correlation:", correlation)
```

# Replace missing ages using NumPy

```python
median_age = np.nanmedian(age)

age_clean = np.where(
    np.isnan(age),
    median_age,
    age
)

print("Missing before:", np.isnan(age).sum())
print("Missing after:", np.isnan(age_clean).sum())
```

# What is numpy and why is it preferred over python lists


NumPy (Numerical Python) is a Python library used for performing numerical, mathematical, and statistical computations efficiently.

NumPy is preferred over Python lists because:
1. **Faster execution**: NumPy performs numerical operations faster, especially on large datasets.
   
2. **Memory efficiency**: NumPy arrays generally consume less memory than Python lists.
   
3. **Vectorized operations**: Mathematical operations can be performed directly on entire arrays without explicit loops.
   
4. **Built-in functions**: NumPy provides functions such as mean(), median(), and std() for statistical analysis.
