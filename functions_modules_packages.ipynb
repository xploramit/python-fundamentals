# Types of arguments

In Python, arguments are commonly explained in four main types:

1. Positional arguments
2. Keyword arguments
3. Default arguments
4. Variable-length arguments

Let's use simple Amazon-like examples throughout.

1. Positional arguments — order matters

Suppose we define:

```python
def product(name, price):
    print(name, price)

product("Laptop", 50000)
```

```python
def product(name, price):
    print(name, price)

product(50000, "Laptop")
```

Python assigns values according to their position:

name  ← "Laptop"
price ← 50000

If you reverse them:

product(50000, "Laptop")

Python takes:

name  ← 50000
price ← "Laptop"

So:
# Positional argument = position/order matters.

# 2. Keyword arguments - name matters

Instead of depending on position, we explicitly mention the parameter name:

```python
def product(name, price):
    print(name, price)

product(price=50000, name="Laptop")
```

Even though price came first, Python understands because we specified:

price = 50000

name = "Laptop"

So:

# Keyword argument = specify which parameter receives which value.

# 3. Default arguments - value is already provided

Suppose standard delivery is the default:

```python
def order(product, delivery="Standard"):
    print(product, delivery)

order("Laptop")
```

We didn't provide a delivery type, so Python used:

delivery = "Standard"

But the customer can choose something else:

```python
order("Laptop", "Express")
```

# Default argument = if the user doesn't provide a value, use the predefined value.

# 4. Variable-length arguments - number of arguments can vary

Sometimes we don't know how many values will be given.

# *args - many positional arguments

Imagine a customer's shopping cart:

# *args - when we don't know how many values will come

Suppose an Amazon-like shopping cart allows a customer to add any number of products.

One customer may buy:

# Laptop

Another may buy:

# Laptop, Mouse, Keyboard

# Another may buy 10 products.

Instead of defining:

def cart(product1, product2, product3, product4):

we can simply write:

```python
def cart(*products):
    print(products)

cart("Laptop", "Mouse", "Keyboard", "Headphones")
```

# So *products says:

"Give me any number of positional arguments. I will collect them together."

Python collects them into a tuple.

We can then process each product:

```python
def cart(*products):
    for product in products:
        print("Added to cart:", product)

cart("Laptop", "Mouse", "Keyboard", "Pen", "Ipad", "Shirt")
```

# This is conceptually similar to an e-commerce cart: the number of products isn't fixed in advance.

# Then what is **kwargs?

Now suppose that after selecting products, the customer provides order details such as:

name = "Amit"

city = "Guwahati"

payment = "UPI"

delivery = "Express"

The number and names of these details may vary.

We can use:

```python
def order(**details):
    print(details)

order(
    name="Amit",
    city="Guwahati",
    payment="UPI",
    delivery="Express"
)
```

# **kwargs collects the information into a dictionary.

So:

# *args → many values without parameter names → tuple
    
# **kwargs → many named/keyword values → dictionary

We can combine both in our Amazon-like example

```python
def amazon_order(*products, **details):
    print("Products:", products)
    print("Order Details:", details)

amazon_order(
    "Laptop",
    "Mouse",
    "Keyboard",
    name="Amit",
    city="Guwahati",
    payment="UPI"
)
```

# * → collects extra positional arguments

# ** → collects extra keyword arguments

Memory trick:

#  🛒 *args = What products did you put in the cart?

#  📋 **kwargs = What named details did you provide about the order?

This is an important exam/interview question.

| Feature                 | `*args`                                   | `**kwargs`                             |
| ----------------------- | ----------------------------------------- | -------------------------------------- |
| **Meaning**             | Accepts multiple **positional arguments** | Accepts multiple **keyword arguments** |
| **Symbol**              | `*`                                       | `**`                                   |
| **Values given as**     | Simple values                             | `key=value` pairs                      |
| **Stored as**           | **Tuple**                                 | **Dictionary**                         |

# What is a Docstring?

A docstring (documentation string) is a description written inside a function to explain what that function does.

Suppose we already have:

```python
discount_rate = 10

def calculate_price(price):
    discount = price * discount_rate / 100
    final_price = price - discount
    return final_price

print(calculate_price(50000))
```

Now imagine another programmer sees:

calculate_price()

They may wonder:

What does this function do? What value should I give it? What will it return?

That's where a docstring helps.

Add a docstring

```python
discount_rate = 10

def calculate_price(price):
    """Calculate the final price after applying the discount."""
    
    discount = price * discount_rate / 100
    final_price = price - discount
    return final_price

print(calculate_price(50000))
```

This part:

"""Calculate the final price after applying the discount."""

is the docstring.

It is not doing the calculation. It is simply explaining what the function does.

# How can we see the docstring?

Python stores the docstring. We can access it using:

# print(calculate_price.__doc__)

```python
discount_rate = 10

def calculate_price(price):
    """
    Calculate the final product price.

    price: Original price of the product.

    Returns:
        Price after applying the discount.
    """
    
    discount = price * discount_rate / 100
    final_price = price - discount
    return final_price


# Show the docstring
print(calculate_price.__doc__)

# Calculate price of ₹50,000 product
print(calculate_price(50000))
```

# .__doc__ → Tell me what this function does.

# calculate_price(50000) → Actually do the calculation.

# Docstring vs Comment

| Docstring                               | Comment                          |
| --------------------------------------- | -------------------------------- |
| Documents a function, class or module   | Explains a piece of code         |
| Usually uses `""" """`                  | Uses `#`                         |
| Python can access it through `.__doc__` | Not available through `.__doc__` |
| Useful for formal documentation         | Useful as notes for programmers  |

# What is an Array?

# An array is a data structure used to store many values together in an ordered form, especially values of the same type.


But we already have Lists. Why do we need Arrays?

Because a Python list can also store these prices:

```python
prices = [50000, 30000, 1000, 2000, 5000]
```

So for a small program, a list is perfectly fine.

The issue arises when we start doing large-scale numerical calculations, data analysis and machine learning.

Suppose an Amazon-like company wants to give a 10% discount to every product.

With a list, this doesn't work the way a beginner might expect:

```python
prices = [1000, 2000, 5000]

print(prices * 2)
```

It repeats the list. It doesn't multiply every price by 2.

To calculate a 10% discount, we'd typically process the elements:

```python
prices = [1000, 2000, 5000]

new_prices = []

for price in prices:
    new_prices.append(price * 0.90)

print(new_prices)
```

# It works, but NumPy arrays are designed specifically for this type of numerical operation.

With an array:

```python
import numpy as np

prices = np.array([1000, 2000, 5000])

new_prices = prices * 0.90

print(new_prices)
```
