# Class and Object

# Class
# A class is a blueprint or template for creating objects.
For example, Amazon sells millions of products.

Instead of separately designing every product, we can create a general blueprint:
# Product
with:
- name
- price
- brand
- rating
This blueprint is called a class.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
```

# Object

An object is an actual instance created from a class.

```python
p1 = Product("iPhone 17", 79999)
p2 = Product("Samsung Galaxy S26", 69999)
```

Here:
Product → Class

p1 → Object

p2 → Object


Class = Blueprint of a house

Object = Actual house constructed from that blueprint

# Attributes

Attributes tell us what an object HAS.

p1 HAS a name → iPhone 17

p1 HAS a price → 79999

p2 HAS a name → Samsung Galaxy S26

p2 HAS a price → 69999

# Let's print the attributes

```python
print(p1.name)
print(p1.price)

print(p2.name)
print(p2.price)
```

# p1 and p2 come from the same class, they contain different data.

# Methods

# Attributes tell us what an object HAS. Methods tell us what an object can DO.

On Amazon, a product might be able to:

display_details()

apply_discount()

check_stock()

# Let's add a method.

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.price)
```

# Now create the objects:

```python
p1 = Product("iPhone 17", 79999)
p2 = Product("Samsung Galaxy S26", 69999)
```

# And call the method:

```python
p1.display_details()
```

```python
p2.display_details()
```

# self

```python
def display_details(self):
    print(self.name)
    print(self.price)
```

“How does Python know whether it should display the iPhone or Samsung?”

Because of self.

When we write:

```python
p1.display_details()
```

# self means "this particular object."

# More Realistic Amazon Method

“Amazon sometimes gives discounts. So let's give our Product objects the ability to calculate a discounted price.”

# Class

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.price)

    def apply_discount(self, discount):
        discount_amount = self.price * discount / 100
        final_price = self.price - discount_amount

        print("Original Price:", self.price)
        print("Discount:", discount, "%")
        print("Final Price:", final_price)
```

# Object

```python
p1 = Product("iPhone 17", 79999)
p2 = Product("Samsung Galaxy S26", 69999)
```

# Method

```python
p1.apply_discount(10)
```

```python
p2.apply_discount(20)
```

Attribute = What an object HAS

Method = What an object DOES

# Constructor (__init__)

The constructor gives initial information to an object when the object is created.

Suppose Amazon creates a new Product object.

It immediately needs information such as:
Name,
Price,
Brand

```python
class Product:

    def __init__(self, name, price, brand):
        self.name = name
        self.price = price
        self.brand = brand
```

```python
p1 = Product("Laptop", 75000, "Lenovo")
```

When p1 is created, __init__ automatically stores:

name = Laptop

price = 75000

brand = Lenovo
