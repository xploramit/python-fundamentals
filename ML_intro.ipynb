# Deep Learning is a subset of Machine Learning, which is a subset of Artificial Intelligence.


# Difference Between Artificial Intelligence (AI), Machine Learning (ML), and Deep Learning (DL)

| Concept | Meaning | Real-World Example |
|---|---|---|
| **Artificial Intelligence (AI)** | Making machines perform tasks that normally require human intelligence. | Amazon Alexa responding to questions. |
| **Machine Learning (ML)** | Machines learn patterns from data to make predictions or decisions. | Netflix recommending movies based on viewing behaviour. |
| **Deep Learning (DL)** | A type of ML that uses neural networks with multiple layers. | Speech recognition or advanced image recognition. |

### Important Note

**Not all AI systems use Machine Learning, and not all Machine Learning uses Deep Learning.**


- **AI:** The broad field of making machines intelligent.
- **ML:** Machines learn patterns from data.
- **DL:** Machines learn through multilayer neural networks.

### Relationship

**Artificial Intelligence (AI) ⊃ Machine Learning (ML) ⊃ Deep Learning (DL)**

This means that **Deep Learning is a subset of Machine Learning, and Machine Learning is a subset of Artificial Intelligence.**

# Overfitting vs Underfitting

These are two important problems in Machine Learning.

**Underfitting**
Underfitting occurs when a model is too simple to capture important patterns in the data.

Example: Netflix recommends the same popular movies to everyone without considering individual preferences.

**Overfitting**
Overfitting occurs when a model learns the training data too closely, including random noise, and performs poorly on new data.

Example: A Netflix recommendation model memorises a user's exact past viewing history but fails to recommend suitable new movies.


## Difference Between Underfitting and Overfitting

| **Basis** | **Underfitting** | **Overfitting** |
|---|---|---|
| **Model** | Too simple. | Too complex for the available data. |
| **Training Performance** | Poor. | Very good. |
| **Testing Performance** | Poor. | Often poor. |
| **Main Problem** | Fails to learn patterns. | Learns noise along with patterns. |
| **Possible Solution** | Use better features or a more flexible model. | Use regularisation, simpler models or more representative data. |

### Important Note

A **good fit** captures useful patterns and performs well on both training and unseen data.


- **Underfitting:** The model has not learned enough.
- **Overfitting:** The model has memorised the training data instead of learning generalisable patterns.
- **Good Fit:** The model learns useful patterns and predicts unseen data accurately.


# Batch Learning vs Online Learning

**Batch Learning**
A model is trained using a collected dataset. When new data becomes available, the model is updated through a new training process.

Example: Amazon trains a demand-forecasting model using historical sales data and updates it every week.

**Online Learning**
A model is updated incrementally as new observations arrive, often without retraining on the entire dataset.

Example: An online shopping platform could update its product recommendation model as new user interactions occur.


## Difference Between Batch Learning and Online Learning

| **Batch Learning** | **Online Learning** |
|---|---|
| Trained on collected batches of data. | Learns incrementally from new data. |
| Updates occur periodically. | Updates can occur continuously or frequently. |
| Often requires more resources for full retraining. | Often efficient for incremental updates. |
| Suitable for relatively stable patterns. | Useful when patterns change rapidly. |

### Important Note

**Online Learning does not mean learning through the internet.** It refers to incremental model updates.


- **Batch Learning:** Learn from a collected dataset and update the model periodically.
- **Online Learning:** Learn incrementally as new data becomes available.

# Types of Machine Learning
Machine Learning is commonly classified into four major types.

# Supervised Learning
Supervised Learning is a type of Machine Learning in which a model learns from labelled data, where the correct output is known.

Example: Amazon has historical information about customers and whether they purchased a product.

**Note:** This is an example of **Supervised Machine Learning (Classification)** because the output variable has two categories: **Yes (1)** and **No (0)**.


# Unsupervised Learning
Unsupervised Learning identifies patterns or groups in data without predefined target labels.

Example: Amazon groups customers according to purchasing behaviour.


## Example: Customer Segmentation Using Unsupervised Learning

| **Customer Group** | **Purchasing Behaviour** |
|---|---|
| Group 1 | Frequently purchases electronics. |

## Supervised vs Unsupervised Learning

| **Basis** | **Supervised Learning** | **Unsupervised Learning** |
|---|---|---|
| **Data** | Labelled data. | No predefined target labels. |
| **Output** | Known during training. | Not supplied as a target. |
| **Objective** | Predict a target variable. | Discover patterns or structures. |
| **Common Tasks** | Regression, classification. | Clustering, dimensionality reduction. |
| **Algorithms** | Linear Regression, Logistic Regression, Decision Trees. | K-Means, DBSCAN, PCA. |
| **Amazon Example** | Predict purchase probability. | Group customers by buying habits. |


- **Supervised Learning:** Learning with known answers (labelled data).
- **Unsupervised Learning:** Discovering hidden patterns without known answers (unlabelled data).

### Important Note

**Supervised Learning** is mainly used for prediction, whereas **Unsupervised Learning** is mainly used to discover patterns, similarities, and structures in data.

### 1. Regression

**Regression** predicts a numerical or continuous value.

**Example:** Amazon wants to predict how much a customer will spend next month.

**Prediction:** ₹5,000.

### 2. Classification

**Classification** predicts a category or class.

**Example:** Amazon wants to predict whether a customer will purchase a product.

**Prediction:** Yes or No.

### Difference Between Regression and Classification

| **Basis** | **Regression** | **Classification** |
|---|---|---|
| **Output** | Numerical value. | Category or class. |
| **Question** | How much? How many? | Which class? Yes or No? |
| **Example** | Predict house price. | Predict loan approval. |
| **Netflix Example** | Predict viewing hours. | Predict whether a user will cancel. |

| **Evaluation Metrics** | MAE, MSE, RMSE, R². | Accuracy, Precision, Recall, F1-score. |


- **Regression:** Predicts numerical values.
- **Classification:** Predicts categories or classes.

### Important Note

**Logistic Regression** is generally used for **Classification**, despite having the word "Regression" in its name.



## Simple Linear Regression vs Multiple Linear Regression

### 1. Simple Linear Regression

**Simple Linear Regression** predicts a numerical dependent variable using **one independent variable**.

**Example:** Predict a student's salary package based on CGPA.

The regression equation is:

$$
\text{Package} = \beta_0 + \beta_1(\text{CGPA}) + \varepsilon
$$

Where:

- $\beta_0$ = Intercept
- $\beta_1$ = Regression coefficient of CGPA
- $\varepsilon$ = Error term

### Example Calculation

Suppose the estimated prediction equation is:

$$
\widehat{\text{Package}} = 0.5 + 0.4(\text{CGPA})
$$

For **CGPA = 8**:

$$
\widehat{\text{Package}} = 0.5 + 0.4(8)
$$

$$
= 0.5 + 3.2 = 3.7
$$

**Interpretation:** The predicted salary package is **₹3.7 lakh per annum (LPA)**.

### Important Note

The regression coefficients used in this example are **hypothetical** and are intended for classroom explanation.



### 2. Multiple Linear Regression

**Multiple Linear Regression** predicts a numerical dependent variable using **two or more independent variables**.

**Example:** Predict salary packages using CGPA, internship experience and interview score.

The regression equation is:

$$
\text{Package} = \beta_0 + \beta_1(\text{CGPA}) + \beta_2(\text{Internship}) + \beta_3(\text{Interview Score}) + \varepsilon
$$

Where:

- $\beta_0$ = Intercept
- $\beta_1, \beta_2, \beta_3$ = Regression coefficients
- $\varepsilon$ = Error term

### Example Calculation

Suppose the estimated prediction equation is:

$$
\widehat{\text{Package}} = 0.5 + 0.3X_1 + 0.4X_2 + 0.05X_3
$$

Where:

- $X_1$ = CGPA
- $X_2$ = Number of internships
- $X_3$ = Interview score out of 10

For **CGPA = 8, Internships = 2, Interview Score = 8**:

$$
\widehat{\text{Package}} = 0.5 + 0.3(8) + 0.4(2) + 0.05(8)
$$

$$
= 0.5 + 2.4 + 0.8 + 0.4
$$

$$
= \boxed{4.1 \text{ lakh per annum}}
$$

**Interpretation:** The predicted salary package is **₹4.1 lakh per annum (LPA)**.

### Comparison: Simple vs Multiple Linear Regression

| **Simple Linear Regression** | **Multiple Linear Regression** |
|---|---|
| One independent variable. | Two or more independent variables. |
| Example: CGPA → Salary. | CGPA + Internship + Interview → Salary. |
| Easier to interpret and visualise. | Captures more predictors. |
| Straight line in two dimensions. | Plane or hyperplane for multiple predictors. |


- **Simple Linear Regression:** One independent variable predicts the dependent variable.
- **Multiple Linear Regression:** Two or more independent variables predict the dependent variable.

### Important Note

The regression coefficients used in this example are **hypothetical** and are intended for classroom explanation.



## Linear Regression vs Logistic Regression

This is an important distinction because both have the word **Regression** in their names, but they generally solve different prediction problems.

### 1. Linear Regression

**Linear Regression** predicts a continuous numerical value.

**Example:** Netflix wants to predict how many hours a customer will watch movies next week.

- **Inputs:** Previous viewing hours, subscription duration and other relevant information.
- **Output:** 12.5 hours.

### 2. Logistic Regression

**Logistic Regression** is primarily used for **Classification**. It estimates the probability of belonging to a class.

**Example:** Netflix wants to predict whether a customer will cancel their subscription.

- **Inputs:** Viewing activity, subscription history and engagement.
- **Output:** 0.85 probability of cancellation (85%).

If the classification threshold is **0.50**, the model predicts that the customer will cancel.

### Logistic Regression Equation

$$
P(Y=1)=\frac{1}{1+e^{-z}}
$$

Where:

$$
z=\beta_0+\beta_1X_1+\cdots+\beta_kX_k
$$

Here:

- $P(Y=1)$ = Probability of belonging to class 1.
- $e$ = Euler's number (approximately 2.718).
- $\beta_0$ = Intercept.
- $\beta_1,\ldots,\beta_k$ = Regression coefficients.
- $X_1,\ldots,X_k$ = Independent variables.

The **Sigmoid Function** transforms a numerical score into a probability between **0 and 1**.


- **Linear Regression:** Predicts numerical values (e.g., 12.5 hours).
- **Logistic Regression:** Predicts probabilities used for classification (e.g., 85% probability of cancellation).

### Important Note

Despite its name, **Logistic Regression is primarily a Classification algorithm**, not an ordinary regression algorithm.



### Linear vs Logistic Regression

| **Basis** | **Linear Regression** | **Logistic Regression** |
|---|---|---|
| **Main Task** | Regression. | Classification. |
| **Output** | Continuous numerical value. | Probability used to predict a class. |
| **Output Range** | Not restricted. | Probability between 0 and 1. |
| **Function** | Linear prediction function. | Sigmoid applied to a linear score. |
| **Netflix Example** | Predict viewing hours. | Predict subscription cancellation. |
| **Amazon Example** | Predict monthly spending. | Predict whether a customer purchases. |
| **Typical Evaluation Metrics** | RMSE, MAE, R². | Accuracy, Precision, Recall, F1-score. |
